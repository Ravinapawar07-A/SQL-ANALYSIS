```python
# Suppose a shop provides discounts based on the purchase amount and membership.
# - Purchase amount must be ₹5,000 or more.
# - If the customer is a member, they receive a 20% discount.
# - Otherwise, they receive a 10% discount.
# - If purchase amount is below ₹5,000, there is no discount.


amount = float(input("Enter your Total Amount : "))

if amount>= 5000:
  membership= input("Are you a Member : yes/ No ")

  if membership=="yes":
    print("Your Discount is 20%")
  else:
    print("Your Discount is 10%")

else:
  print ("No discount available ")


```

```python
# Nested if – Practice 1
# Easy

# Question: Take a number from the user.
# 1. First, check whether the number is positive.
# 2. If it is positive, check whether it is even or odd.
# 3. If it is negative or zero, print an appropriate message.

num=int(input ("Enter the number"))

if num>0 :
  print("Positive Number")

  if num%2==0:
   print("It is even number")

  else:
    print("it is a odd number")
elif num<0:
  print("It is a negative number")
else:
  print("Number is zero")
```

```python
# Write a nested if program:
# 1. Take a number from the user.
# 2. If the number is positive, check whether it is divisible by 5.
# 3. If divisible by 5, print "Divisible by 5".
# 4. Otherwise, print "Not divisible by 5".
# 5. If the number is negative, print "Negative number".
# 6. If it is zero, print "Zero".

num= int(input("Enter your number : "))

if num>0:
 print ("it is a positive number")

 if num%5==0:
  print("It is divisible by 5")

 else:
    print("not divisible by 5")
elif num<0:
  print("Its a negative number")

else:
  print("the number is Zero")
```

```python
# Create a student result system using nested if.

# 1. Take marks as input.
# 2. If marks are 35 or above:
#    - Check whether marks are 75 or above.
#    - If yes, print "Distinction".
#    - Otherwise, print "Pass".
# 3. If marks are below 35, print "Fail".

marks= int(input("Enter your marks: "))

if marks>=35:

  if marks>=75:
    print("Distintion")

  else:
    print("pass")
elif marks<0:
    print ("Invalid marks")
else:
  print("fail")
```
